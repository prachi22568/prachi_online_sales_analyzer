## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import numpy as np
import pandas as pd
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_20_dataset.csv")
display(df)
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
print("Shape of dataset:", df.shape)
print("Columns:", df.columns)
print("Dataset information:")
df.info()
print("Missing values:")
print(df.isnull().sum())
print("Number of duplicate rows:", df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
quantity_array = df["Quantity"].to_numpy()
price_array = df["Unit_Price"].to_numpy()

print("Average quantity:", np.mean(quantity_array))
print("Maximum quantity:", np.max(quantity_array))
print("Minimum quantity:", np.min(quantity_array))

print("Average unit price:", np.mean(price_array))
print("Maximum unit price:", np.max(price_array))
print("Minimum unit price:", np.min(price_array))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
df["Revenue"] = df["Quantity"] * df["Unit_Price"]

# Create another useful column using NumPy
df["Order_Value_Level"] = np.where(df["Revenue"] >= 500, "High Value", "Regular")

# Show orders with revenue of 500 or more
high_value_orders = df[df["Revenue"] >= 500]
print("Orders with revenue of 500 or more:")
display(high_value_orders)

# Sort orders from highest to lowest revenue
sorted_orders = df.sort_values(by="Revenue", ascending=False)
print("Top five orders by revenue:")
display(sorted_orders.head())
```

## 6. Your individual project challenge
**Challenge:** Calculate revenue; identify best-selling product and category-wise revenue; rank products.

Write your solution below and explain what the result means.


```python
# Summarise quantity sold and revenue for each product
product_summary = df.groupby("Product").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Number_of_Orders=("Order_ID", "count")
).reset_index()

# Rank products by total quantity sold (highest quantity gets rank 1)
product_summary = product_summary.sort_values(
    by=["Total_Quantity", "Total_Revenue"],
    ascending=[False, False]
).reset_index(drop=True)
product_summary.insert(0, "Rank", range(1, len(product_summary) + 1))

print("Product ranking:")
display(product_summary)

# Calculate category-wise revenue
category_revenue = df.groupby("Category").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Quantity=("Quantity", "sum")
).reset_index().sort_values(by="Total_Revenue", ascending=False)

print("Category-wise revenue:")
display(category_revenue)

# Identify the best-selling product by total quantity sold
best_selling_product = product_summary.iloc[0]
print("Best-selling product:", best_selling_product["Product"])
print("Quantity sold:", best_selling_product["Total_Quantity"])

# Calculate total revenue from all orders
print("Total store revenue:", df["Revenue"].sum())
```

## 7. Export your report
Use `to_csv(..., index=False)` to save your cleaned or analysed DataFrame. Create an `output` folder if needed.


```python
df.to_csv("Student_20_order_analysis.csv", index=False)

# Save product ranking and category revenue summaries
product_summary.to_csv("Student_20_product_ranking.csv", index=False)
category_revenue.to_csv("Student_20_category_revenue.csv", index=False)

print("CSV files saved successfully.")
display(df)
```
